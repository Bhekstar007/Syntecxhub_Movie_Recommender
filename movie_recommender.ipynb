Step 1: Import relevant libraries

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv('TMDB_movie_dataset_v11.csv')
```

Step 4: Filter dataset to a more manageable size

```python
filtered = data[
    (data['status'] == 'Released') &
    (data['vote_count'] >= 50) &
    (data['overview'].notna()) &
    (data['genres'].notna())
].copy()

print(filtered.shape)
```

Step 5: Reset and select relevant columns

```python
filtered = filtered.reset_index(drop=True)

movies = filtered[['id', 'title', 'genres', 'overview', 'tagline', 'vote_average', 'vote_count']].copy()
movies.head(10)
```

Step 6: Handle any missing values in the tagline, overview and genres columns

```python
movies['tagline'] = movies['tagline'].fillna('')
movies['overview'] = movies['overview'].fillna('')
movies['genres'] = movies['genres'].fillna('')

movies.isnull().sum()
```

Step 7: Build a combined content feature

```python
movies['content'] = movies['genres'] + ' ' + movies['overview'] + ' ' + movies['tagline']
movies[['title', 'content']].head()
```

Step 8: Use td-idf to vectorize combined content

```python
from sklearn.feature_extraction.text import TfidfVectorizer

vectorizer = TfidfVectorizer(stop_words='english', max_features=10000)
tfidf_matrix = vectorizer.fit_transform(movies['content'])

print(tfidf_matrix.shape)
```

Step 9: Calculate cosine similarity based on previous vectors

```python
from sklearn.metrics.pairwise import cosine_similarity

cosine_sim = cosine_similarity(tfidf_matrix, tfidf_matrix)

print(cosine_sim.shape)
```

Step 10: Build recommender function

```python
# Build a mapping from movie title to its row index
indices = pd.Series(movies.index, index=movies['title']).drop_duplicates()

def recommend(title, n=10):
    if title not in indices:
        return f"'{title}' not found in the dataset."
    
    idx = indices[title]
    
    # Get similarity scores for this movie against all others
    sim_scores = list(enumerate(cosine_sim[idx]))
    
    # Sort by similarity score, descending
    sim_scores = sorted(sim_scores, key=lambda x: x[1], reverse=True)
    
    # Skip the first result (it's the movie itself, similarity = 1.0)
    sim_scores = sim_scores[1:n+1]
    
    movie_indices = [i[0] for i in sim_scores]
    
    return movies[['title', 'genres', 'vote_average']].iloc[movie_indices]
```

Step 11: Test with sample queries

```python
recommend('Inception')
```

Qualitative Evaluation

Testing the recommender on several sample movies showed a mix of strong and weaker results. Franchise-based queries (e.g. "The Dark Knight", "Ice Age") produced excellent recommendations, almost entirely surfacing other films in the same franchise. This is likely because shared vocabulary (character names, franchise terms) in the overview/tagline text reinforced genre similarity.

Standalone films with more unique thematic content (e.g. "Inception") produced recommendations that matched on genre but didn't fully capture the film's distinctive narrative concepts, since those concepts aren't always explicitly stated in the overview text. Occasional unrelated matches also appeared (e.g. a horror film recommended alongside Iron Man), reflecting a known limitation of content-based filtering: it can only detect similarity based on the literal text provided, not deeper semantic or thematic meaning.

Overall, the recommender performs well for franchise and genre-consistent queries, with room for improvement on more nuanced, standalone films.

Step 12: Flask API Deployment

To demonstrate the recommender as a deployable service rather than just a notebook, the model was wrapped in a Flask API with a single /recommend endpoint. It accepts a title query parameter and optional n for number of results and returns JSON-formatted recommendations.

Example request: http://127.0.0.1:5000/recommend?title=The%20Dark%20Knight&n=5

Testing confirmed the API correctly reproduces the notebook's recommendations, successfully returning results for both "Inception" and "The Dark Knight" queries.

Conclusion

This project built a content-based movie recommendation system using the Full TMDB Movies Dataset 2024. Starting from nearly 1 million movie records, the dataset was filtered down to 27,844 released movies with sufficient vote counts and complete overview/genre metadata. This was a necessary step, since computing pairwise similarity across the full dataset might have been difficult to do.

Genre, overview, and tagline text were combined into a single content field per movie, then converted into numeric features using TF-IDF vectorization. Cosine similarity was computed across all movies, and a recommendation function was built to return the top N most similar titles for any given movie.

Qualitative evaluation: Across several sample queries showed strong performance on franchise-based films like "The Dark Knight" and "Ice Age" both returned highly relevant sequels and spin-offs, while standalone films with more distinctive narrative concepts like "Inception" for example produced recommendations that matched on genre but didn't always capture deeper thematic similarity. This is an expected limitation of content-based filtering using only genre, overview, and tagline text, since the model can only detect similarity present in the literal text provided, without access to keywords or cast/crew data that could strengthen thematic matching.

Flask extension: the recommender was packaged as a Flask API with a /recommend endpoint, accepting a movie title and returning JSON-formatted recommendations. This demonstrates the model functioning as a deployable service rather than only within a notebook environment, successfully reproducing the same recommendations available in the core analysis.

Limitations and future improvements:
1. Incorporating cast, crew, or keyword data which was not available in this dataset could improve thematic matching for standalone films
2. Using n-grams or weighting overview text more heavily could help capture multi-word concepts currently diluted by single-word TF-IDF features
3. A hybrid approach combining content-based filtering with collaborative filtering using user rating data could address cases where content alone isn't sufficient to capture what makes two movies genuinely similar to viewers
4. The Flask API is currently a local development server only; production deployment would require a proper WSGI server and likely a pre-computed, saved similarity matrix rather than recalculating it on every startup

Overall, this project demonstrates a functional, well-evaluated content-based recommendation system, with an honest assessment of where it performs strongly  and where it performed weakly, along with a working deployment as a proof of concept.
